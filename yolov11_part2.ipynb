```
import cv2
import os
from google.colab.patches import cv2_imshow
from ultralytics import YOLO

# Load YOLO model and set to evaluation mode
model_path = "/content/best.pt"
assert os.path.exists(model_path), "❌ Model file not found!"
model = YOLO(model_path)
model.eval()  # Ensure YOLO is in evaluation mode

# Confidence Threshold
conf_threshold = 0.5

# List of image paths
image_paths = ["/content/1007_png.rf.9d8fc7dd299b4201b12d7350c71591b4.jpg"]  # Add more images if needed

# Output directory
output_dir = "/content/labelled_images"
os.makedirs(output_dir, exist_ok=True)

for image_path in image_paths:
    print(f"🔍 Processing: {image_path}")

    # Load Image
    image = cv2.imread(image_path)
    assert image is not None, f"❌ Error reading {image_path}"

    # Resize image to match model input
    image_resized = cv2.resize(image, (640, 640))

    # Run Inference
    results = model(image_resized)

    # Draw bounding boxes
    fish_count = 0
    for result in results:
        boxes = result.boxes.xyxy  # Bounding box coordinates
        confidences = result.boxes.conf  # Confidence scores
        classes = result.boxes.cls  # Class IDs

        # Apply confidence threshold
        keep = confidences > conf_threshold  # Filter out low-confidence detections
        boxes = boxes[keep]
        confidences = confidences[keep]
        classes = classes[keep]

        fish_count = len(boxes)  # Count the number of detected fish

        for box, conf, cls in zip(boxes, confidences, classes):
            x1, y1, x2, y2 = map(int, box)  # Convert to integers
            confidence = float(conf)  # Convert to float
            class_id = int(cls)  # Convert to int (assuming only "fish")

            # Define class name
            class_name = "fish"  # Change if needed

            # Draw bounding box
            cv2.rectangle(image_resized, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Label format: "fish (score)"
            label = f"{class_name} ({confidence:.2f})"
            cv2.putText(image_resized, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX,
                        0.5, (0, 255, 0), 2, cv2.LINE_AA)

    # Overlay fish count text
    cv2.putText(image_resized, f"Fish Count: {fish_count}", (30, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2, cv2.LINE_AA)

    # Show Output Image
    cv2_imshow(image_resized)

    # Save the image
    output_path = os.path.join(output_dir, os.path.basename(image_path))
    cv2.imwrite(output_path, image_resized)
    print(f"✅ Image saved: {output_path}")

print("🎯 Processing complete! Images saved in:", output_dir)
```
